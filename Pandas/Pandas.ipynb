## loc & iloc
.loc[] selects data by exact column names (labels), whereas .iloc[] selects data by integer positions (index numbers).

````python
#loc[row, column]
print(coffee_csv.loc[2])
print("\n",coffee_csv.loc[2:, "Day"])
print("\n",coffee_csv.loc[:5, ["Day","Coffee Type"]])
````

* `df.fillna(value)` fills missing values (`NaN`) with a specified default or aggregated statistic (e.g., `.mean()`).

## .agg()
(short for aggregate) lets you apply one or multiple summary functions (like sum, mean, count, min, max, or custom functions) to a DataFrame, Series, or GroupBy object all at once.

````python
coffee_csv.groupby('Coffee Type').agg({'Units Sold' : 'sum', 'price' : 'mean'})
````

## pivot()

* .pivot() only reshapes your DataFrame—it moves existing values into a grid without doing any math, so every (index, column) pair must be unique.
* .pivot_table() aggregates (does math) AND reshapes—it combines duplicate pairs using an aggregation function (like sum, mean, or count) before placing the result into the grid.



````python
pivote = coffee_csv.pivot_table(index = 'Day', columns='Coffee Type', values='Total Revenue', aggfunc ='sum')
print("\n")
print(coffee_csv.pivot_table(index = 'Day', columns='Coffee Type', values='Total Revenue', aggfunc ='mean'))
````

````python
pivote['Espresso']
````

````python
# pivote['Day'] == 'Thursday'
# will raise a KeyError: 'Day' because 'Day' is no longer a regular column—you made it the index when you set index='Day' in your pivot.
pivote.loc['Thursday']
````

* pivote.sum(axis=1) → Adds across columns → Gives total revenue per Day (Monday, Tuesday, etc.).

* pivote.sum(axis=0) → Adds down rows → Gives total revenue per Coffee Type (Espresso, Latte, etc.).



````python
print("Adds down rows\n\n",pivote.sum()) #by default axis = 0
print("\n")
print("Adds across columns\n\n",pivote.sum(axis=1))
````
